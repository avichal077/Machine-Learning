benifits

1. each machine startes with a random chace of being the best
2. every time you play, you update what you know about each machine
3. instead of just picking the machine with the highest average, you ramdomly choose based on how confident you are that a machine is the best

why is it better
even if the machine was lucky early, thompson sampling still tests other machines in a smart way. it doesn't get stuck on a bad choice

```python
# random module random number generate karta hai 

import random

#step 1: number of slot machine 
num_machines = 3

#step 2: total trials
n_trials = 1000

#step 3: actual winning probabilities
true_rewards = [0.2, 0.5, 0.8]

#step 4: success counter
# har machine kitni baar jeeti
# intially kisi machine ne kuch nahi jeeta
success=[0]*num_machines

#step 5: failure counter
# har machine kitni baar haari
failure=[0]*num_machines

#main learning loop
for trial in range (n_trials):
    #step 6: sample probability from the beta distribution
    # har machine ke liye ek probabilty sample hogi
    # beta(success+1, failure+1)
    sampled_values = [random.betavariate(success[machine]+1, failure[machine]+1) for machine in range(num_machines)]

    #step 7: choose machine
    # jis machine ka sampled probability sabse jyada hogi woh choose karenge
    machine = sampled_values.index(max(sampled_values))

    #step 8: check reward
    #environment reward generate karega
    # example
    #machine3 probability 0.8 hai, toh agar random number 0.45 hai reward milega
    reward = 1 if random.random() < true_rewards[machine] else 0

    #step 9: update knowledge
    # agar reward mila

    if reward == 1:
        # success count increment hoga
        success[machine] += 1

    else:
        # failure count increment hoga
        failure[machine] += 1

#final results
print("\n final results\n")
for machine in range(num_machines):
    #total kitni baar machine kheli gayi
    total_played=success[machine]+failure[machine]

    #agar machine kabhi choose hi nahi hui
    if total_played==0:
        print(f"Machine {machine+1}: never played")
    else:
        #estimated win rate
        estimated_rate=success[machine]/total_played
        print(f"machine {machine+1}: played {total_played} times, estimated win rate: {estimated_rate:.2f}, true win rate: {true_rewards[machine]:.2f}")

#best machine

#jis machine ke sabse jyada wins hai
best_machine = success.index(max(success))
print(f"\nBest machine is machine {best_machine+1} with estimated win rate: {success[best_machine]/(success[best_machine]+failure[best_machine]):.2f} and true win rate: {true_rewards[best_machine]:.2f}")


```
